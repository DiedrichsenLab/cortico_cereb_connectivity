### This notebook loads the f-contrast maps from the RETROICOR regressors and plot them

```python
import numpy as np
import nibabel as nib
from nilearn import plotting
import matplotlib.pyplot as plt
import os
```

#### F-contrast maps of RETROICOR

```python
fmap_path = '/cifs/diedrichsen/data/Cerebellum/Social/data/GLM_physio/glm_task0_retro1_hr0'
subj_list = ['sub-05']

for sub in subj_list:
    file_path = f'{fmap_path}/{sub}/spmF_0001.nii'
    plotting.plot_glass_brain(file_path, display_mode='lyrz', colorbar=True,
                            plot_abs=False, title=f'{sub}')
    plotting.show()
```

#### T-contrast maps of HR*CRF when task+HR regressors in GLM

```python
map_path = '/cifs/diedrichsen/data/Cerebellum/Social/data/GLM_physio/glm_task1_retro0_hr1'
anat_path = '/cifs/diedrichsen/data/Cerebellum/Social/data/anatomicals/'
subj_list = ['sub-05', 'sub-07', 'sub-08', 'sub-09', 'sub-11', 'sub-12', 'sub-13', 'sub-16', 
'sub-17', 'sub-18', 'sub-19', 'sub-20', 'sub-21', 'sub-22', 'sub-25', 'sub-27']

for sub in subj_list:
    anat_img = nib.load(f'{anat_path}/{sub}/anatomical.nii')
    
    display = plotting.plot_stat_map(
            f'{map_path}/{sub}/spmT_0001.nii',
            bg_img=anat_img,
            display_mode="ortho",
            cut_coords=(0,0,0),
            threshold=5.0,
            colorbar=True,
            title=f"{sub}"
        )
    plt.show()
```
